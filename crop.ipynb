```python
fig, axes = plt.subplots(2, 2, figsize=(14,12))
#line plot 
lab = agg.groupby("label")["ph"].mean().head(5)
sns.lineplot(x=lab.index, y=lab.values,marker="*",ax=axes[0,0])
axes[0,0].set_xlabel("Label")
axes[0,0].set_ylabel("Average pH")
axes[0,0].set_title("Average pH by Label")
#Histplot
rain = agg.groupby("humidity")["N"].mean()
sns.histplot(x=rain.index,bins=10,ax=axes[0,1],palette="coolwarm" )
axes[0,1].set_xlabel("Humidity")
axes[0,1].set_ylabel("Average N")
axes[0,1].set_title("Histoplot")
#pie chart
lab=agg.groupby("label")["temperature"].mean().nlargest(5)
axes[1,0].pie(
    lab.values,
    labels=lab.index,
    autopct="%1.1f%%",
    startangle=90
)
axes[1,0].set_title("Temperature by Label")
#countplot
sns.heatmap(agg.isnull(),ax=axes[1,1])

```

```python
fig, axes = plt.subplots(2, 2, figsize=(14,12))
#barplot
rain = agg.groupby("label")["rainfall"].mean().head(5)
sns.barplot(x=rain.index, y=rain.values,palette="coolwarm",ax=axes[0,0])
axes[0,0].set_xlabel("Label")
axes[0,0].set_ylabel("Average Rainfall")
axes[0,0].set_title("Average Rainfall by Label")

#scatter plot
plt.figure(figsize=(7,4))
sns.scatterplot(data=agg, x="temperature", y="rainfall",size="ph",ax=axes[0,1])
axes[0,1].set_title("Temperature vs Rainfall")
axes[0,1].set_xlabel("temperature")
axes[0,1].set_ylabel("rainfall")

# barplot
top10 = agg["label"].value_counts().nlargest(10)
sns.barplot(x=top10.index,y=top10.values,ax=axes[1,0],palette="pastel")
axes[1,0].set_xlabel("Crop")
axes[1,0].set_ylabel("Count")
axes[1,0].set_title("Top 10 Crop Count")
axes[1,0].tick_params(axis="x", rotation=45)

#hisplot curve
sns.histplot(data=agg, x="temperature", bins=10,kde=True,ax=axes[1,1],color="purple")
axes[1,1].set_title("Temperature Distribution")



```

```python
fig, axes = plt.subplots(2, 2, figsize=(18,16))
#pointplot
sns.pointplot(data=agg,x="label",y="temperature",ax=axes[0,0])
axes[0,0].tick_params(axis="x", rotation=45)
axes[0,0].set_title("Temperature by Crop Label")
axes[0,0].set_xlabel("Crop Label")
axes[0,0].set_ylabel("Temperature")
#boxplot
sns.boxplot(data=agg,x="label",y="humidity",color="green",ax=axes[0,1])
axes[0,1].tick_params(axis="x", rotation=45)
#violineplot
top10 = agg.nlargest(10, "temperature")
sns.violinplot(data=top10,x="label",y="temperature",ax=axes[1,0])
axes[1,0].set_xlabel("Label")
axes[1,0].set_ylabel("Temperature")
axes[1,0].set_title("Temperature Distribution by Label")
#barplot
top10=agg.nlargest(5,"ph")[["rainfall","label","ph"]]
sns.barplot(data=top10,x="rainfall",y="ph",orient="h",ax=axes[1,1],color="purple",alpha=0.5)
axes[1,1].set_title("")


```
